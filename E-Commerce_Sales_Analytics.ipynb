### 2. Sales

```python
df_sales.info()
df_sales
```

vamos alterar a tipagem de *'order_date'* e *'order_time'*

```python
df_sales['order_datetime']=pd.to_datetime((df_sales['order_date'] + ' ' + df_sales['order_time']),format='%Y-%m-%d %H:%M:%S')
df_sales=df_sales.drop(columns=['order_date','order_time'])
df_sales
```

As features abaixo possuem valores nulos, vamos avaliar.
* delivery_days             
* estimated_delivery_days   
* return_status             
* return_reason 
* customer_rating 
* review_sentiment 
* customer_review 
* campaign_name             
* coupon_code   

```python
print(df_sales.shape)
df_sales.isna().sum()
```

```python
display(df_sales.groupby('delivery_status')['delivery_days'].size())

display(df_sales.groupby('delivery_status')['estimated_delivery_days'].size())
```

Pelas tabelas acima, vemos que os registros do 'delivery_status'=='Cancelled' tem a mesma quantidade de registro para dados nulos para as features *delivery_days* e *estimated_delivery_days*. Podemos simplesmente deletar o status de Cancelled do df.
   

```python
df_sales=df_sales[df_sales['delivery_status']!='Cancelled']
df_sales.info()
```

Com isso já resolvemos boa parte do nosso df. Vamos olhar agora para as features *return_status* e *return_reason*. As colunas não tem nenhum registro, vamos tirar do df.

```python
df_sales=df_sales.drop(columns=['return_status','return_reason'])
df_sales.info()
```

Vamos nos concentrar agora com *'campaign_name'* e '*coupon_code*'

```python
df_sales['campaign_name'].value_counts()
```

Para campaign_name vamos preencher os registros null como no campaign. Algo semelhante vamos fazer para coupon_code.

```python
df_sales['campaign_name']=df_sales['campaign_name'].fillna('no campaign')
df_sales['coupon_code']=df_sales['coupon_code'].fillna('no coupon')
df_sales.info()
```

Agora está tudo correto 😁
